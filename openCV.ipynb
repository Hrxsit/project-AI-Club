```python
import cv2
from matplotlib.pyplot import imread
import numpy as np

img = cv2.imread("C:/Users/airfo/PycharmProjects/first/cat.jpg",cv2.IMREAD_COLOR)

# RESIZE
img = cv2.resize(img,(200,400))
img = cv2.resize(img,(0,0),fx=2,fy=1)
#it will make dimensions as [800,800]

#CROP
height , width = img.shape[0] , img.shape[1]
img=img[int(height/4):,50:-50]

#ROTATE
img=cv2.rotate(img,cv2.ROTATE_180)
#now rotating at any angle
M=cv2.getRotationMatrix2D((width/2,height/2),150,1)
#it is in the form center(x,y),rotation '+' is counterclockwise and scale(i.e magnification)
#it only gives a matrix . it dosen't actually rotates.
img=cv2.warpAffine(img,M,(2*width,int(1.2*height)))
#it is in the form image to be rotated , matrix , dimensions of the window.

#TRANSLATE
fx=width/5
fy=-height/5
M=np.array([
    [1,0,fx]
    ,[0,1,fy]
])
img=cv2.warpAffine(img,M,(width,height))

cv2.imshow("Cat",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
img = cv2.imread("C:/Users/airfo/PycharmProjects/first/cat.jpg",cv2.IMREAD_COLOR)


# BORDER
img = cv2.copyMakeBorder(img, top=50, bottom=50, left=50, right=50, borderType=cv2.BORDER_CONSTANT,
                        value=(100, 0, 0))

# LINE
img = cv2.line(img, pt1=(700, 675), pt2=(700, 550), color=(0, 200, 0), thickness=20)
img = cv2.line(img, (625, 675), (700, 550), color=(0, 200, 0), thickness=20)
img = cv2.line(img, (225, 675), (300, 800), color=(0, 200, 0), thickness=20)
img = cv2.line(img, (225, 675), (400, 750), color=(0, 200, 0), thickness=20)

# ARROW
img = cv2.arrowedLine(img, pt1=(50, 50), pt2=(450, 600), color=(0, 0, 200),thickness=10)

# CIRCLE // center, color, radius, thickness
img = cv2.circle(img, center=(1450, 800), color=(100, 100, 0),
                radius=100, thickness=-1)

# ELLIPSE // center, axes, angle, startAngle, endAngle, color, thickness
img = cv2.ellipse(img, center=(300, 685), axes=(50, 30),
                angle=100, startAngle=0, endAngle=270,
                color=(0, 200, 200), thickness=20)

# RECTANGLE
img = cv2.rectangle(img, pt1=(50, 1200), pt2=(2000, 1400),
                    color=(150, 150, 150), thickness=-1)

# TEXT
img = cv2.putText(img, "ELLA <3", org=(200, 300),
                fontFace=cv2.FONT_HERSHEY_SIMPLEX, fontScale=1,
                color=(0, 0, 0), thickness=1)

cv2.imshow("ELLA", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2

# Replace "0" with a file path to work with a saved video
stream = cv2.VideoCapture(0)

if not stream.isOpened():
    print("No stream :(")
    exit()

fps = stream.get(cv2.CAP_PROP_FPS)
if fps == 0 or fps is None:
    fps = 30.0  # Fallback to a standard webcam frame rate
width = int(stream.get(3))
height = int(stream.get(4))

# list of FourCC video codes: https://softron.zendesk.com/hc/en-us/articles/207695697-List-of-FourCC-codes-for-video-codecs
output = cv2.VideoWriter("C:/Users/airfo/PycharmProjects/first/stream.mp4",
            cv2.VideoWriter_fourcc('m', 'p', '4', 'v'),
            fps=fps, frameSize=(width, height))

while True:
    ret, frame = stream.read()
    if not ret: # if no frames are returned
        print("No more stream :(")
        break

    frame = cv2.resize(frame, (width, height))
    output.write(frame)
    cv2.imshow("Webcam!", frame)
    if cv2.waitKey(1) == ord('q'): # press "q" to quit
        break

stream.release()
output.release()
cv2.destroyAllWindows() #!
```

```python
import cv2

# Open webcam
stream = cv2.VideoCapture(0)

if not stream.isOpened():
    print("No stream :(")
    exit()

# Fix 1: Handle potential 0.0 FPS from webcam
fps = stream.get(cv2.CAP_PROP_FPS)
if fps <= 0:
    fps = 30.0  # Safe fallback standard

width = int(stream.get(cv2.CAP_PROP_FRAME_WIDTH))   # Explicitly using property name
height = int(stream.get(cv2.CAP_PROP_FRAME_HEIGHT)) # Explicitly using property name

# Define codec and create VideoWriter object
# Note: 'mp4v' works beautifully with .mp4 containers
output = cv2.VideoWriter("C:/Users/airfo/PycharmProjects/first/stream.mp4",
                cv2.VideoWriter_fourcc(*'mp4v'), # Clean shorter syntax for ('m','p','4','v')
                fps=fps,frameSize=(width, height))

print(f"Recording started at {width}x{height} @ {fps} FPS. Press 'q' to stop...")

while True:
    ret, frame = stream.read()
    if not ret:
        print("No more stream :(")
        break

    # Write the frame to the file
    output.write(frame)

    # Display the live preview
    cv2.imshow("Webcam!", frame)

    if cv2.waitKey(1) == ord('q'):
        break

# Clean up everything properly
stream.release()
output.release()  # Fix 2: Explicitly save and close the video file
cv2.destroyAllWindows()
print("Video saved successfully!")
```

```python
import cv2

face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades +
                    "haarcascade_frontalface_default.xml")
smile_cascade = cv2.CascadeClassifier(cv2.data.haarcascades +
                    "haarcascade_smile.xml")
eye_cascade = cv2.CascadeClassifier(cv2.data.haarcascades +
                    "haarcascade_eye.xml")

def detect_features(frame):
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    faces = face_cascade.detectMultiScale(gray, 1.3, 5)
            # You can also specify minSize and maxSize
    for (x, y, w, h) in faces:
        frame = cv2.rectangle(frame, (x, y), (x+w, y+h),
                            color=(0, 255, 0), thickness=5)
        face = frame[y : y+h, x : x+w]
        gray_face = gray[y : y+h, x : x+w]
        smiles = smile_cascade.detectMultiScale(gray_face,
                            3.0, minNeighbors=4)
        for (xp, yp, wp, hp) in smiles:
            face = cv2.rectangle(face, (xp, yp), (xp+wp, yp+hp),
                    color=(0, 0, 255), thickness=3)

        eyes = eye_cascade.detectMultiScale(gray_face,
                    3.0, minNeighbors=3)
        for (xp, yp, wp, hp) in eyes:
            face = cv2.rectangle(face, (xp, yp), (xp+wp, yp+hp),
                    color=(255, 0, 0), thickness=3)

    return frame

stream = cv2.VideoCapture(0)

if not stream.isOpened():
    print("No stream :(")
    exit()

fps = stream.get(cv2.CAP_PROP_FPS)
width = int(stream.get(3))
height = int(stream.get(4))

# list of FourCC video codes: https://softron.zendesk.com/hc/en-us/articles/207695697-List-of-FourCC-codes-for-video-codecs
output = cv2.VideoWriter("C:/Users/airfo/PycharmProjects/first/face_stream.mp4",
            cv2.VideoWriter_fourcc('m', 'p', '4', 'v'),
            fps=fps, frameSize=(width, height))

while(True):
    ret, frame = stream.read()
    if not ret:
        print("No more stream :(")
        break

    frame = detect_features(frame)
    output.write(frame)
    cv2.imshow("Webcam!", frame)
    if cv2.waitKey(1) == ord('q'):
        break

stream.release()
output.release()
cv2.destroyAllWindows() #!
```
